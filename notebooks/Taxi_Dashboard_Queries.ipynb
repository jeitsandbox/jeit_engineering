```python
# Cell: Dashboard Query 1 - Top 10 Busiest Zones
print("📊 TOP 10 BUSIEST PICKUP ZONES")
print("="*60)
display(spark.sql("""
    SELECT 
        zone_name,
        borough,
        taxi_type,
        trip_count,
        CONCAT('$', FORMAT_NUMBER(total_revenue, 2)) as revenue
    FROM jeit_kg_dev.nyc_tlc.vw_top_pickup_zones
    ORDER BY trip_count DESC
    LIMIT 10
"""))
```

```python
# Cell: Dashboard Query 2 - Hourly Trip Pattern by Time of Day
print("📊 HOURLY TRIP PATTERNS")
print("="*60)
display(spark.sql("""
    SELECT 
        pickup_hour,
        time_of_day,
        is_rush_hour,
        SUM(total_trips) as trips,
        ROUND(AVG(avg_fare), 2) as avg_fare
    FROM jeit_kg_dev.nyc_tlc.vw_peak_hours
    WHERE pickup_year = 2024
    GROUP BY pickup_hour, time_of_day, is_rush_hour
    ORDER BY pickup_hour
"""))
```

```python
# Cell: Dashboard Query 3 - Green vs Yellow Comparison
print("📊 GREEN VS YELLOW TAXI COMPARISON")
print("="*60)
display(spark.sql("""
    SELECT 
        taxi_type,
        pickup_year,
        total_trips,
        avg_distance,
        avg_duration,
        avg_fare,
        CONCAT('$', FORMAT_NUMBER(total_revenue, 2)) as revenue
    FROM jeit_kg_dev.nyc_tlc.vw_green_vs_yellow
    ORDER BY taxi_type, pickup_year
"""))
```

```python
# Cell: Dashboard Query 4 - Payment Method Preferences
print("📊 PAYMENT METHOD ANALYSIS")
print("="*60)
display(spark.sql("""
    SELECT 
        taxi_type,
        payment_type_desc,
        SUM(trip_count) as total_trips,
        ROUND(AVG(avg_fare), 2) as avg_fare,
        ROUND(AVG(avg_tip_percentage), 2) as avg_tip_pct,
        CONCAT('$', FORMAT_NUMBER(SUM(total_revenue), 2)) as revenue
    FROM jeit_kg_dev.nyc_tlc.agg_payment_analysis
    GROUP BY taxi_type, payment_type_desc
    ORDER BY total_trips DESC
"""))
```

```python
# Cell: Dashboard Query 5 - Day of Week Patterns
print("📊 BUSIEST DAYS OF THE WEEK")
print("="*60)
display(spark.sql("""
    SELECT 
        day_of_week_desc,
        taxi_type,
        SUM(total_trips) as trips,
        ROUND(AVG(avg_fare), 2) as avg_fare
    FROM jeit_kg_dev.nyc_tlc.vw_day_of_week_patterns
    WHERE pickup_year = 2024
    GROUP BY day_of_week_desc, pickup_day_of_week, taxi_type
    ORDER BY pickup_day_of_week, taxi_type
"""))
```

```python
# Cell: Dashboard Query 6 - Monthly Revenue Trends
print("📊 MONTHLY REVENUE TRENDS")
print("="*60)
display(spark.sql("""
    SELECT 
        pickup_year,
        pickup_month,
        taxi_type,
        total_trips,
        CONCAT('$', FORMAT_NUMBER(total_revenue, 2)) as revenue
    FROM jeit_kg_dev.nyc_tlc.vw_monthly_trends
    ORDER BY pickup_year, pickup_month, taxi_type
"""))
```
